```python
import pandas as pd
import numpy as np
```

```python
relatorio = pd.read_csv("Relatorio Tecnologia.csv", encoding="cp1252")
relatorio = relatorio.rename(columns={'DIA TRABALHADO:': 'DATA'})
relatorio['DATA'] = pd.to_datetime(relatorio['DATA'], format='%d/%m/%Y')
display(relatorio)
```

```python
relatorio['DATA'] = pd.to_datetime(relatorio['DATA'], format='%d/%m/%Y')
data_unica = relatorio['DATA'].drop_duplicates().sort_values(ascending=True).reset_index(drop=True)

df_trimestre = pd.DataFrame({
    'data': data_unica.dt.strftime('%d-%m-%Y'),
    'mes': data_unica.dt.month,
    'trimestre': data_unica.dt.quarter,
})
display(df_trimestre)
```

```python
df_primeiro_trimestre = df_trimestre[df_trimestre['trimestre'] == 1]
display(df_primeiro_trimestre)

df_segundo_trimestre = df_trimestre[df_trimestre['trimestre'] == 2]
display(df_segundo_trimestre)

df_terceiro_trimestre = df_trimestre[df_trimestre['trimestre'] == 3]
display(df_terceiro_trimestre)
```

```python
# display('Valores nulos')
# display(relatorio.isna())
relatorio.fillna('Sem ocorrências', inplace=True)
display(relatorio)
```

```python
# Cria uma nova coluna no dataframe:
# relatorio['TRIMESTRE'] = relatorio['DATA'].dt.quarter 
relatorio.insert(2, 'TRIMESTRE', relatorio['DATA'].dt.quarter)
display(relatorio)
```

```python
relatorio['QTD_FALHAS'] = (relatorio == 'NÃO CONFORME').sum(axis=1)
display(relatorio)
```

```python
falhas_trimestres = relatorio.groupby('TRIMESTRE')['QTD_FALHAS'].sum()
display(falhas_trimestres)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(falhas_trimestres.index, falhas_trimestres.values, color='skyblue')

plt.title('Grafico de ocorrências de cada trimestre', fontsize=14, fontweight='bold')
plt.xlabel('Trimestre', fontsize=12)
plt.ylabel('Quantidade de falhas', fontsize=12)

plt.xticks(falhas_trimestres.index)

plt.show()
```

```python
# Lista com o nome dos equipamentos
equipamentos = ['MOCHILINKS', 'RETORNO VIVOS', 'MULTIVIEW3', 'MULTIVIEW2', 'GLOBOPLAY', 'MULTIVIEW', 'CLEARCOM']

tabela_equipamentos = relatorio.groupby('TRIMESTRE')[equipamentos].apply(lambda x: (x == 'NÃO CONFORME').sum())

display(tabela_equipamentos)
```

```python
tabela_equipamentos.plot(kind='bar', figsize=(10, 6))
```

```python
ax = tabela_equipamentos.plot(kind='bar', figsize=(10, 6), edgecolor='black')

plt.title('Ocorrências de "Não Conforme" por Equipamento em Cada Trimestre', fontsize=14, fontweight='bold')
plt.xlabel('Equipamentos', fontsize=12)
plt.ylabel('Quantidade de Falhas', fontsize=12)

plt.xticks(rotation=360, ha='right')

plt.legend(title='Trimestre', title_fontsize='11', fontsize='10')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
ax = tabela_equipamentos.plot(kind='line', marker='o', linewidth=2.5, markersize=8, figsize=(10, 6))

plt.xticks([1,2,3], ['1º', '2º', '3º'], fontsize=11)
plt.yticks(range(0,7))
plt.title('Ocorrências de cada equipamento')
plt.ylabel('Quantidade de falhas', fontsize=12)
plt.legend(title='Equipamentos', title_fontsize=11, fontsize=10, bbox_to_anchor=(1.02,1), loc='upper left')
plt.grid(axis='y', linestyle='--', alpha=0.7)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.savefig('slide4_evolucao.png', dpi=300, bbox_inches='tight', transparent=True)

plt.tight_layout()
plt.show()
```

```python
color= ["#a377ee", "#41b3fa","#f38b46"]

ax = tabela_equipamentos.T.plot(kind='bar', stacked=True, figsize=(10,6), edgecolor='black', width=0.6, color=color)

plt.xticks(rotation=45, ha='right', fontsize=10)
plt.legend(['1º Trimestre', '2º Trimestre', '3º Trimestre'], title='Período', title_fontsize=11, fontsize=10, bbox_to_anchor=(1.02,1), loc='upper left')
plt.title('Ocorrências de cada equipamento')
plt.ylabel('Quantidade de falhas', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
for container in ax.containers:
    labels = [int(v) if v > 0 else '' for v in container.datavalues]
    ax.bar_label(container, labels=labels, label_type='center', color='white', fontweight='bold', fontsize=10)

plt.savefig('slide3_empilhado.png', dpi=300, bbox_inches='tight', transparent=True)

plt.tight_layout()
plt.show()
```

```python
tabela_familias = pd.DataFrame({
    'Retorno Vivos': tabela_equipamentos['RETORNO VIVOS'],
    'Família Multiview': tabela_equipamentos['MULTIVIEW'] + tabela_equipamentos['MULTIVIEW2'] + tabela_equipamentos['MULTIVIEW3'],
    'Mochilinks': tabela_equipamentos['MOCHILINKS'],
    'Clearcom': tabela_equipamentos['CLEARCOM'],
    'Globoplay': tabela_equipamentos['GLOBOPLAY']
})

# Soma os 3 trimestres de cada família e ordena do menor para o maior (para a maior barra ficar no topo)
totais_familias = tabela_familias.sum().sort_values(ascending=True)

cores = ["#6eda5e", "#e4f153", "#ba75f6", '#f28e2b', '#e15759']

ax = totais_familias.plot(kind='barh', figsize=(10, 5), color=cores, edgecolor='black', width=0.6)

plt.grid(axis='x', linestyle='--', alpha=0.7)
ax.bar_label(ax.containers[0], padding=5, fontweight='bold')
plt.xlim(0, 11.5)

plt.title('Total de Ocorrências "Não Conforme" por Família de Equipamento', fontsize=14, fontweight='bold')
plt.xlabel('Total de Falhas Acumuladas (1º ao 3º Trimestre)', fontsize=12)
plt.ylabel('')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.savefig('slide2_familias.png', dpi=300, bbox_inches='tight', transparent=True)

plt.tight_layout()
plt.show()
```
