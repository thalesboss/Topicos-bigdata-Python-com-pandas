```python
import pandas as pd
import numpy as np
```

```python
relatorio = pd.read_csv("Relatorio Tecnologia.csv", encoding="cp1252")
relatorio = relatorio.rename(columns={'DIA TRABALHADO:': 'DATA'})
relatorio['DATA'] = pd.to_datetime(relatorio['DATA'], format='%d/%m/%Y')
display(relatorio)
```

```python
relatorio['DATA'] = pd.to_datetime(relatorio['DATA'], format='%d/%m/%Y')
data_unica = relatorio['DATA'].drop_duplicates().sort_values(ascending=True).reset_index(drop=True)

df_trimestre = pd.DataFrame({
    'data': data_unica.dt.strftime('%d-%m-%Y'),
    'mes': data_unica.dt.month,
    'trimestre': data_unica.dt.quarter,
})
display(df_trimestre)
```

```python
df_primeiro_trimestre = df_trimestre[df_trimestre['trimestre'] == 1]
display(df_primeiro_trimestre)

df_segundo_trimestre = df_trimestre[df_trimestre['trimestre'] == 2]
display(df_segundo_trimestre)

df_terceiro_trimestre = df_trimestre[df_trimestre['trimestre'] == 3]
display(df_terceiro_trimestre)
```

```python
# display('Valores nulos')
# display(relatorio.isna())
relatorio.fillna('Sem ocorrências', inplace=True)
display(relatorio)
```

```python
# Cria uma nova coluna no dataframe:
# relatorio['TRIMESTRE'] = relatorio['DATA'].dt.quarter 
relatorio.insert(2, 'TRIMESTRE', relatorio['DATA'].dt.quarter)
display(relatorio)
```

```python
relatorio['QTD_FALHAS'] = (relatorio == 'NÃO CONFORME').sum(axis=1)
display(relatorio)
```

```python
falhas_trimestres = relatorio.groupby('TRIMESTRE')['QTD_FALHAS'].sum()
display(falhas_trimestres)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(falhas_trimestres.index, falhas_trimestres.values, color='skyblue')

plt.title('Grafico de ocorrências de cada trimestre', fontsize=14, fontweight='bold')
plt.xlabel('Trimestre', fontsize=12)
plt.ylabel('Quantidade de falhas', fontsize=12)

plt.xticks(falhas_trimestres.index)

plt.show()
```

```python
# Lista com o nome dos equipamentos
equipamentos = ['MOCHILINKS', 'RETORNO VIVOS', 'MULTIVIEW3', 'MULTIVIEW2', 'GLOBOPLAY', 'MULTIVIEW', 'CLEARCOM']

tabela_equipamentos = relatorio.groupby('TRIMESTRE')[equipamentos].apply(lambda x: (x == 'NÃO CONFORME').sum())

display(tabela_equipamentos)
```

```python
tabela_equipamentos.plot(kind='bar', figsize=(10, 6))
```

```python
ax = tabela_equipamentos.plot(kind='bar', figsize=(10, 6), edgecolor='black')

plt.title('Ocorrências de "Não Conforme" por Equipamento em Cada Trimestre', fontsize=14, fontweight='bold')
plt.xlabel('Equipamentos', fontsize=12)
plt.ylabel('Quantidade de Falhas', fontsize=12)

plt.xticks(rotation=360, ha='right')

plt.legend(title='Trimestre', title_fontsize='11', fontsize='10')

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
